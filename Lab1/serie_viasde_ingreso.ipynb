```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from statsmodels.tsa.seasonal import seasonal_decompose

pd.set_option("display.max_columns" , None)

archivo = Path( "Laboratorio 1. Series de Tiempo 2026 - Base_Migracion_2009-2026jun.xlsx" )

datos = pd.read_excel(archivo)

print("dataset:", datos.shape)
print(datos.columns.tolist())
```

```python
datos["Año"] = pd.to_numeric( datos[ "Año"], errors ="coerce")
datos["Mes cod"] = pd.to_numeric(datos["Mes cod"], errors ="coerce")
datos["Viajero"] = pd.to_numeric( datos["Viajero"] , errors ="coerce")

datos["Fecha"] = pd.to_datetime(
    {"year": datos["Año"], "month": datos["Mes cod"], "day": 1},

    errors ="coerce"
)
```

```python
tipos_seleccionados = ["Turista", "Excursionista"]

datos_visitantes = datos[datos["Tipo de Viajero"].isin(tipos_seleccionados)].copy()

datos["Tipo de Viajero"].isin(tipos_seleccionados)
```

```python
datos_visitantes["Vía"].value_counts(dropna=False)
```

```python
series_vias = (
    datos_visitantes
    .dropna(subset=["Fecha", "Vía", "Viajero"])
    .groupby(["Fecha", "Vía"])["Viajero"]
    .sum()
    .unstack("Vía")
    .sort_index()
    .asfreq("MS")
)
```

### SARIMA

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.statespace.sarimax import SARIMAX

modelos_aerea = [
    ((0, 1, 0), (0, 0, 0, 12)),
    ((1, 1, 0), (0, 0, 0, 12)),
    ((0, 1, 1), (0, 0, 0, 12)),
    ((1, 1, 1), (0, 0, 0, 12)),
    ((1, 1, 0), (1, 0, 0, 12)),
    ((0, 1, 1), (0, 0, 1, 12)),
]
```

```python
resultados_aerea = []

for orden, orden_estacional in modelos_aerea:

    modelo = SARIMAX(
        log_aerea,
        order=orden,
        seasonal_order=orden_estacional,
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    ajuste = modelo.fit(disp = False)

    prediccion_log = ajuste.get_forecast(steps= len(prueba_aerea )).predicted_mean

    prediccion = np.exp(prediccion_log )
    prediccion.index = prueba_aerea.index

    mae = mean_absolute_error( prueba_aerea, prediccion )

    rmse = np.sqrt(
        mean_squared_error(
            prueba_aerea,
            prediccion
        )
    )

    resultados_aerea.append({
        "orden": orden,
        "orden_estacional": orden_estacional,
        "aic": ajuste.aic,
        "bic": ajuste.bic,
        "mae": mae,
        "rmse": rmse
    })

comparacion_aerea = pd.DataFrame(resultados_aerea )

comparacion_aerea =(
    comparacion_aerea
    .sort_values("rmse" )
    .reset_index(drop = True )
)

comparacion_aerea
```

```python
modelo_aerea_final = SARIMAX(
    log_aerea,
    order = (0 , 1 , 1),
    seasonal_order= (0 , 0, 1, 1 ),
    enforce_stationarity = False ,
    enforce_invertibility = False
)

ajuste_aerea_final = modelo_aerea_final.fit( isp = False)

print(ajuste_aerea_final.summary() )
```

### Ljung-Box:

```python
from statsmodels.stats.diagnostic import acorr_ljungbox

residuos_aerea = ajuste_aerea_final.resid.dropna()

inicio_residuos = ajuste_aerea_final.loglikelihood_burn
residuos_aerea = residuos_aerea.iloc[inicio_residuos:]

ljung_box_aerea = acorr_ljungbox(
    residuos_aerea,
    lags= [12, 24 ],
    model_df= 2 ,
    return_df= True
)

ljung_box_aerea
```

```python
modelo_aerea_2 = SARIMAX(
    log_aerea,
    order = (0 ,1 ,1),
    seasonal_order= (0 , 0, 0 , 12),
    enforce_stationarity = False ,
    enforce_invertibility = False
)

ajuste_aerea_2 = modelo_aerea_2.fit(disp= False )

print(ajuste_aerea_2.summary())
```

```python
modelo_aerea_3 = SARIMAX(
    log_aerea,
    order=(1, 1 , 0 ),
    seasonal_order=(0, 0, 0, 12 ),
    enforce_stationarity= False ,
    enforce_invertibility=False
)

ajuste_aerea_3 = modelo_aerea_3.fit(disp= False)

print(ajuste_aerea_3.summary() )
```

```python
def revisar_ljung_box(ajuste, grados_modelo):

    residuos = ajuste.resid.dropna()
    inicio = ajuste.loglikelihood_burn
    residuos = residuos.iloc[inicio: ]

    return acorr_ljungbox(
        residuos,
        lags= [12, 24 ],
        model_df=grados_modelo ,
        return_df= True
    )
```

```python
print( "Modelo MA(1):" )
display(revisar_ljung_box(ajuste_aerea_2, 1) )
 
print( "Modelo AR(1):")
display(revisar_ljung_box(ajuste_aerea_3, 1 ))
```

### Seasonal Naive

```python
def pronostico_naive_estacional(serie_entrenamiento, pasos, periodo=12):
    valores_estacionales = serie_entrenamiento.iloc[-periodo:].to_numpy()

    repeticiones = int(np.ceil(pasos / periodo))

    pronostico = np.tile(
        valores_estacionales,
        repeticiones
    )[:pasos]

    return pronostico
```

#### aérea

```python
prediccion_naive_aerea = pronostico_naive_estacional(
    entrenamiento_aerea ,
    len(prueba_aerea )
)

mae_naive_aerea = mean_absolute_error( 
    prueba_aerea ,
    prediccion_naive_aerea

)

rmse_naive_aerea = np.sqrt(

    mean_squared_error(
        prueba_aerea ,
        prediccion_naive_aerea
    )
)

print("MAE aérea:", mae_naive_aerea )
print("RMSE aérea:", rmse_naive_aerea )
```

#### terrestre

```python
prediccion_naive_terrestre = pronostico_naive_estacional( entrenamiento_terrestre, len(prueba_terrestre) )

mae_naive_terrestre = mean_absolute_error(
    prueba_terrestre,
    prediccion_naive_terrestre
)

rmse_naive_terrestre = np.sqrt(

    mean_squared_error(
        prueba_terrestre,
        prediccion_naive_terrestre )
)

print("MAE terrestre:", mae_naive_terrestre )
print("RMSE terrestre:", rmse_naive_terrestre )
```
