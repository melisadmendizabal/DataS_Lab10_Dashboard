```python
def escalar_serie(entrenamiento, prueba):

    escalador = MinMaxScaler(feature_range = (0 , 1) )

    valores_entrenamiento = entrenamiento.to_numpy().reshape(-1 , 1 )

    valores_prueba = prueba.to_numpy().reshape(-1 , 1 )

    entrenamiento_escalado = escalador.fit_transform(valores_entrenamiento)

    prueba_escalada = escalador.transform(valores_prueba)

    return entrenamiento_escalado, prueba_escalada, escalador
```

```python
def crear_ventanas(valores, longitud_ventana):
    entradas = []
    salidas = []

    for posicion in range(
        longitud_ventana,
        len(valores)
    ):
        entradas.append(valores[ posicion - longitud_ventana: posicion])

        salidas.append(valores[posicion])

    return np.array(entradas), np.array(salidas)
```

```python
def crear_ventanas(valores, longitud_ventana):
    entradas = []
    salidas = []

    for posicion in range(
        longitud_ventana,
        len(valores)
    ):
        entradas.append(valores[posicion - longitud_ventana: posicion] )

        salidas.append(valores[posicion])

    return np.array(entradas) , np.array(salidas)
```

```python
def preparar_datos_lstm(
    entrenamiento,
    prueba,
    longitud_ventana,
    proporcion_validacion=0.20
):
    (
        entrenamiento_escalado,
        prueba_escalada,
        escalador
    ) = escalar_serie(entrenamiento,prueba)

    x_completo, y_completo = crear_ventanas(entrenamiento_escalado,longitud_ventana)

    cantidad_validacion = int(len(x_completo) * proporcion_validacion)

    posicion_corte = (len(x_completo) - cantidad_validacion)

    x_entrenamiento = x_completo[:posicion_corte]
    y_entrenamiento = y_completo[:posicion_corte]

    x_validacion = x_completo[posicion_corte:]
    y_validacion = y_completo[posicion_corte:]

    return {
        "x_entrenamiento": x_entrenamiento,
        "y_entrenamiento": y_entrenamiento,
        "x_validacion": x_validacion,
        "y_validacion": y_validacion,
        "entrenamiento_escalado": entrenamiento_escalado,
        "prueba_escalada": prueba_escalada,
        "escalador": escalador
    }
```

```python
def construir_lstm_simple(
    cantidad_unidades,
    tasa_dropout,
    tasa_aprendizaje,
    longitud_ventana
):
    modelo = Sequential([
        Input(shape=(longitud_ventana, 1)),

        LSTM(units=cantidad_unidades),

        Dropout(rate=tasa_dropout),

        Dense(1)
    ])

    modelo.compile(
        optimizer = Adam(learning_rate = tasa_aprendizaje),
        loss="mse",
        metrics=["mae"]
    )

    return modelo
```

```python
def construir_lstm_apilada(
    unidades_primera_capa,
    unidades_segunda_capa,
    tasa_dropout,
    tasa_aprendizaje,
    longitud_ventana
):
    modelo = Sequential([
        Input(shape= (longitud_ventana, 1)),

        LSTM( units=unidades_primera_capa,
            return_sequences=True
        ),

        Dropout(rate = tasa_dropout),

        LSTM(units= unidades_segunda_capa),

        Dropout(rate= tasa_dropout),

        Dense(1)
    ])

    modelo.compile(
        optimizer = Adam(learning_rate=tasa_aprendizaje),
        loss="mse",
        metrics=["mae"]
    )

    return modelo
```

```python
from itertools import product
```

```python
def tunear_modelos_lstm(
    nombre_serie,
    entrenamiento,
    prueba
):
    configuraciones = []

    ventanas = [6, 12, 18]
    unidades = [16, 32, 64]
    tasas_dropout = [0.0, 0.2]
    tasas_aprendizaje = [0.001, 0.0005]
    tamanos_batch = [8, 16]

    parada_temprana = EarlyStopping(
        monitor="val_loss",
        patience=15,
        restore_best_weights=True
    )

    
    for (
        longitud_ventana,
        cantidad_unidades,
        tasa_dropout,
        tasa_aprendizaje,
        tamano_batch
    ) in product(
        ventanas,
        unidades,
        tasas_dropout,
        tasas_aprendizaje,
        tamanos_batch
    ):
        datos = preparar_datos_lstm(
            entrenamiento,
            prueba,
            longitud_ventana
        )

        tf.keras.backend.clear_session()
        tf.random.set_seed(semilla)

        modelo = construir_lstm_simple(
            cantidad_unidades=cantidad_unidades,
            tasa_dropout=tasa_dropout,
            tasa_aprendizaje=tasa_aprendizaje,
            longitud_ventana=longitud_ventana
        )

        historial = modelo.fit(
            datos["x_entrenamiento"],
            datos["y_entrenamiento"],
            validation_data=(
                datos["x_validacion"],
                datos["y_validacion"]
            ),
            epochs=150,
            batch_size=tamano_batch,
            callbacks=[parada_temprana],
            shuffle=False,
            verbose=0
        )

        mejor_val_loss = min(historial.history["val_loss"] )

        configuraciones.append({
            "serie": nombre_serie,
            "arquitectura": "LSTM simple",
            "ventana": longitud_ventana,
            "unidades_1": cantidad_unidades,
            "unidades_2": None,
            "dropout": tasa_dropout,
            "learning_rate": tasa_aprendizaje,
            "batch_size": tamano_batch,
            "epocas_utilizadas": len(
                historial.history["loss"]
            ),
            "val_loss": mejor_val_loss
        })

 
    unidades_primera = [32, 64]
    unidades_segunda = [16, 32]

    for (
        longitud_ventana,
        unidades_1,
        unidades_2,
        tasa_dropout,
        tasa_aprendizaje,
        tamano_batch
    ) in product(
        ventanas,
        unidades_primera,
        unidades_segunda,
        tasas_dropout,
        tasas_aprendizaje,
        tamanos_batch
    ):
        datos = preparar_datos_lstm(
            entrenamiento,
            prueba,
            longitud_ventana
        )

        tf.keras.backend.clear_session()
        tf.random.set_seed(semilla)

        modelo = construir_lstm_apilada(
            unidades_primera_capa=unidades_1,
            unidades_segunda_capa=unidades_2,
            tasa_dropout=tasa_dropout,
            tasa_aprendizaje=tasa_aprendizaje,
            longitud_ventana=longitud_ventana
        )

        historial = modelo.fit(
            datos["x_entrenamiento"],
            datos["y_entrenamiento"],
            validation_data=(
                datos["x_validacion"],
                datos["y_validacion"]
            ),
            epochs=150,
            batch_size=tamano_batch,
            callbacks=[parada_temprana],
            shuffle=False,
            verbose=0
        )

        mejor_val_loss = min(historial.history["val_loss"] )

        configuraciones.append({
            "serie": nombre_serie,
            "arquitectura": "LSTM apilada",
            "ventana": longitud_ventana,
            "unidades_1": unidades_1,
            "unidades_2": unidades_2,
            "dropout": tasa_dropout,
            "learning_rate": tasa_aprendizaje,
            "batch_size": tamano_batch,
            "epocas_utilizadas": len(
                historial.history["loss"]
            ),
            "val_loss": mejor_val_loss
        })

    resultados = pd.DataFrame(configuraciones)

    return resultados.sort_values(
        "val_loss"
    ).reset_index(drop=True)
```

```python
def tunear_lstm_reducido( nombre_serie, entrenamiento , prueba ):
    resultados = []

    configuraciones = [
        {
            "arquitectura": "LSTM simple",
            "ventana": 6,
            "unidades_1": 16,
            "unidades_2": None,
            "dropout": 0.2,
            "learning_rate": 0.001,
            "batch_size": 16
        },

        {
            "arquitectura": "LSTM simple",
            "ventana": 12,
            "unidades_1": 32 ,
            "unidades_2": None ,
            "dropout": 0.0,
            "learning_rate": 0.0005,
            "batch_size": 8
        } ,

        {
            "arquitectura": "LSTM apilada",
            "ventana": 6,
            "unidades_1": 32 ,
            "unidades_2": 16 ,
            "dropout": 0.0 , 
            "learning_rate": 0.0005 ,
            "batch_size": 8
        } ,

        {
            "arquitectura": "LSTM apilada",
            "ventana": 12,
            "unidades_1": 32,
            "unidades_2": 16 ,
            "dropout": 0.2 ,
            "learning_rate": 0.001 ,
            "batch_size": 16
        }
    ]

    parada_temprana = EarlyStopping(
        monitor = "val_loss" ,
        patience = 12 ,
        restore_best_weights=True
    )

    for configuracion in configuraciones:

        datos = preparar_datos_lstm(
            entrenamiento ,
            prueba ,
            configuracion["ventana" ]
        )

        tf.keras.backend.clear_session()
        tf.random.set_seed(semilla)

        if configuracion["arquitectura" ] == "LSTM simple":

            modelo = construir_lstm_simple(
                cantidad_unidades=configuracion["unidades_1"],
                tasa_dropout=configuracion["dropout"],
                tasa_aprendizaje=configuracion["learning_rate"],
                longitud_ventana=configuracion["ventana"]
            )

        else:
            modelo = construir_lstm_apilada(
                unidades_primera_capa= configuracion["unidades_1" ],
                unidades_segunda_capa= configuracion["unidades_2" ],
                tasa_dropout= configuracion["dropout" ],
                tasa_aprendizaje= configuracion["learning_rate" ],
                longitud_ventana= configuracion["ventana" ]
            )

        historial = modelo.fit(
            datos["x_entrenamiento"],
            datos["y_entrenamiento"],
            validation_data = ( datos["x_validacion"],  datos["y_validacion"] ),
            epochs = 100,
            batch_size= configuracion[ "batch_size"] ,
            callbacks = [parada_temprana],
            shuffle = False,
            verbose = 0
        )

        resultados.append({
            "serie": nombre_serie,
            **configuracion,
            "epocas_utilizadas": len( historial.history["loss"] ),
            "val_loss": min( historial.history[ "val_loss"] )
        })

    return (
        pd.DataFrame(resultados)
        .sort_values("val_loss")
        .reset_index(drop=True)
    )
```

```python
def entrenar_y_predecir_lstm(
    entrenamiento ,
    prueba ,
    longitud_ventana,
    unidades,
    dropout,
    learning_rate ,
    batch_size,
    semilla = 42
):

    
    tf.keras.backend.clear_session()
    np.random.seed(semilla)
    tf.random.set_seed(semilla)

    escalador = MinMaxScaler(feature_range = (0 , 1) )

    entrenamiento_valores = ( entrenamiento.to_numpy().reshape( -1 , 1) )

    entrenamiento_escalado = escalador.fit_transform( entrenamiento_valores )

    x_completo, y_completo = crear_ventanas( entrenamiento_escalado, longitud_ventana )

    cantidad_validacion = int( len(x_completo) * 0.20 )

    posicion_corte = ( len(x_completo) - cantidad_validacion )

    x_entrenamiento = x_completo[:posicion_corte]
    y_entrenamiento = y_completo[:posicion_corte]

    x_validacion = x_completo[posicion_corte:]
    y_validacion = y_completo[posicion_corte:]

    
    modelo = construir_lstm_simple(
        cantidad_unidades = unidades ,
        tasa_dropout = dropout,
        tasa_aprendizaje = learning_rate ,
        longitud_ventana = longitud_ventana
    )

    parada_temprana = EarlyStopping(
        monitor= "val_loss",
        patience = 15 ,
        restore_best_weights = True
    )

    historial = modelo.fit(
        x_entrenamiento,
        y_entrenamiento,
        validation_data = ( x_validacion, y_validacion ),
        epochs = 150 ,
        batch_size = batch_size,
        callbacks = [parada_temprana] ,
        shuffle = False,
        verbose = 0
    )


    ventana_actual = entrenamiento_escalado[ -longitud_ventana: ].copy()

    predicciones_escaladas = []

    for _ in range(len(prueba)):

        entrada_modelo = ventana_actual.reshape( 1, longitud_ventana, 1)

        siguiente_valor = modelo.predict(
            entrada_modelo,
            verbose = 0
        )[ 0 , 0]

        predicciones_escaladas.append( siguiente_valor )

        ventana_actual = np.vstack([
            ventana_actual[1:] ,
            [ [siguiente_valor] ]
        ] )

    
    predicciones = escalador.inverse_transform(
        np.array( predicciones_escaladas ).reshape( -1 , 1)
    ).flatten()

    predicciones = pd.Series(
        predicciones,
        index =prueba.index ,
        name = "prediccion_lstm"
    )

    
    mae = mean_absolute_error( prueba, predicciones )

    rmse = np.sqrt( mean_squared_error( prueba, predicciones))

    return {
        "modelo": modelo ,
        "historial": historial,
        "predicciones": predicciones ,
        "mae": mae,
        "rmse": rmse,
        "epocas": len(historial.history[ "loss" ]),
        "escalador": escalador
    }
```
